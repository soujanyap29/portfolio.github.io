# Neural–Symbolic Knowledge Base Construction System

This notebook presents a publication-style end-to-end implementation that transforms raw text into a structured knowledge graph with neural extraction and symbolic refinement.

## 1) System Architecture
Frontend (Streamlit) → Backend Interface → Preprocessing → Neural Extraction (Ollama) → Triplet Parsing → Confidence Estimation → Symbolic Reasoning → Graph Construction → Storage → Inference → Querying → Evaluation

## 2) Module Descriptions
- Preprocessing module cleans and segments input text.
- Neural extractor sends strict prompts to a local Ollama model.
- Triplet parser validates `(subject, relation, object)` structure.
- Confidence estimator assigns and filters triples.
- Symbolic engine removes vague/invalid relations, normalizes entities, and deduplicates.
- Graph builder stores directional edges with relation/confidence.
- Storage manager persists graph JSON.
- Inference engine performs multi-hop reasoning.
- Query engine supports subject/object/relation retrieval.
- Evaluation module reports quality and consistency metrics.

## 5) Results and Observations
- Raw triples are generated by the neural extractor.
- Confidence filtering removes weak triples.
- Symbolic refinement improves consistency and deduplicates entities.
- Inference expands knowledge through multi-hop connections.

## 6) Evaluation Discussion
The evaluation module tracks raw vs accepted/rejected/refined triples and computes simple stability indicators (validity ratio and consistency score) useful for iterative prompt and rule improvements.

## 7) Limitations
- Extraction quality depends on local model behavior and prompt adherence.
- Lightweight confidence scoring is heuristic.
- Type constraints are rule-based and can be extended for domain-specific ontologies.

## 8) Conclusion
This implementation demonstrates an integrated neural–symbolic pipeline from raw text to queryable and inferable knowledge graphs, with a notebook backend and Streamlit frontend interface suitable for research prototyping and extension.
