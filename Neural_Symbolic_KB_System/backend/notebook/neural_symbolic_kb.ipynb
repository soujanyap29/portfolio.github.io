# Neural–Symbolic Knowledge Base Construction System

This notebook demonstrates the full end-to-end neural–symbolic pipeline from raw text to knowledge graph construction, inference, querying, and evaluation.

## Architecture
Frontend (Streamlit) → Backend API → Preprocessing → Neural Extraction (Ollama/LLaMA3) → Parsing → Confidence → Symbolic Reasoning → Graph Building → Storage → Inference → Query → Evaluation

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import networkx as nx
import pandas as pd

PROJECT_ROOT = Path.cwd().parents[1]
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from backend.api_interface import process_text, get_triples, get_graph, query_graph
```

## Input and Pipeline Execution

```python
input_text = '''
Alice works at Acme Corp.
Acme Corp is located in Bangalore City.
Bangalore City is in India Country.
'''
results = process_text(input_text)
results
```

## Triple Outputs

```python
triples = get_triples()
raw_df = pd.DataFrame(triples['raw_triples'])
filtered_df = pd.DataFrame(triples['filtered_triples'])
refined_df = pd.DataFrame(triples['refined_triples'])
inferred_df = pd.DataFrame(triples['inferred_triples'])
raw_df, filtered_df, refined_df, inferred_df
```

## Querying

```python
query_graph('relation: located in')
```

## Knowledge Graph Visualization

```python
graph_payload = get_graph()
G = nx.node_link_graph(graph_payload, directed=True)
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_size=1800, font_size=8, arrows=True)
edge_labels = {(u, v): d.get('relation', '') for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=7)
plt.title('Neural–Symbolic Knowledge Graph')
plt.axis('off')
plt.show()
```
